# Preprocessing of the Language Files

## Loading of the Training files

```python
import pandas as pd

DATA_PATH = "../data/raw/"
AFR_TR = "afr_train.csv"
DEU_TR = "deu_train.csv"
RON_TR = "ron_train.csv"

with open(DATA_PATH + AFR_TR, "r") as afr_tr:
    afr_tr_df = pd.read_csv(afr_tr)
    print("Afrikaans Training Data:")
    print(afr_tr_df.head(3))

with open(DATA_PATH + DEU_TR, "r") as deu_tr:
    deu_tr_df = pd.read_csv(deu_tr)
    print("German Training Data:")
    print(deu_tr_df.head(3))

with open(DATA_PATH + RON_TR, "r") as ron_tr:   
    ron_tr_df = pd.read_csv(ron_tr)
    print("Romanian Training Data:")
    print(ron_tr_df.head(3))
```

## Unused Column Removal

```python
afr_tr_df = afr_tr_df.drop(columns=["id"])
deu_tr_df = deu_tr_df.drop(columns=["id"])
ron_tr_df = ron_tr_df.drop(columns=["id"])

print(afr_tr_df.head(1))
print(deu_tr_df.head(1))
print(ron_tr_df.head(1))
```

## Safe Files in Processed

```python
afr_tr_df.to_csv("../data/processed/p_afr_train.csv", index=False)
deu_tr_df.to_csv("../data/processed/p_deu_train.csv", index=False)
ron_tr_df.to_csv("../data/processed/p_ron_train.csv", index=False)
```
